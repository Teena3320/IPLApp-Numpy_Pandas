SETUP: Initial Setup

```python
import numpy as np

data = np.genfromtxt(
    'deliveries.csv',
    delimiter=',',
    dtype=None,
    encoding='utf-8',
    names=True
)
print(data.dtype.names)

match_id = data['match_id']
batting_team = data['batting_team']
bowling_team = data['bowling_team']
over = data['over']
ball = data['ball']
batter = data['batter']
bowler = data['bowler']
batsman_runs = data['batsman_runs']
total_runs = data['total_runs']
```

Task 1: Total Runs Per Match

```python
unique_matches, match_idx = np.unique(match_id, return_inverse=True)

runs_per_match = np.bincount(match_idx, weights=total_runs)

task1_result = np.column_stack((unique_matches, runs_per_match))

print("{:<10} {:>10}".format("Match ID", "Total Runs"))
print("-" * 21)

for mid, runs in task1_result[:10]:
    print("{:<10} {:>9}".format(int(mid), int(runs)))
```

Task 2: Top 5 Batters

```python
unique_batters, batter_idx = np.unique(batter, return_inverse=True)

runs_per_batter = np.bincount(batter_idx, weights=batsman_runs)

top5_idx = np.argsort(runs_per_batter)[-5:][::-1]

top5_batters = unique_batters[top5_idx]
top5_runs = runs_per_batter[top5_idx]

task2_result = list(zip(top5_batters, top5_runs))
print("Top 5 Batters by Total Runs")
print("-" * 27)

for rank, (player, runs) in enumerate(task2_result, start=1):
    print(f"{rank}. {player:<12}       {int(runs)}")
```

Task 3: Strike Rate of Batters

```python
balls_per_batter = np.bincount(batter_idx)
strike_rate = np.where(balls_per_batter > 0, (runs_per_batter / balls_per_batter) * 100, 0)

task3_result= list(zip(unique_batters, strike_rate))
print("Strike Rate of Each Batter")
print("-" * 27)

for player, sr in task3_result:
    print(f"{player:<18}  {sr:.2f}")
```

Task 4: Economy Rate of Bowlers

```python
unique_bowlers, bowler_idx = np.unique(bowler, return_inverse=True)
runs_conceded_per_bowler = np.bincount(bowler_idx, weights=total_runs)

overs_per_bowler = np.bincount(bowler_idx)
min_overs=50
balls_threshold = min_overs * 6

economy_rate = np.where(overs_per_bowler > balls_threshold, (runs_conceded_per_bowler / (overs_per_bowler/6)), 0)
task4_result = list(zip(unique_bowlers, economy_rate))
print("Economy Rate of Each Bowler")
print("-" * 27)

for player, er in task4_result:
    if er > 0:
        print(f"{player:<18}   {er:.2f}")
```

Task 5: Runs per Over

```python
unique_overs, over_idx = np.unique(over, return_inverse=True)
runs_per_over = np.bincount(over_idx, weights=total_runs)
balls_per_over = np.bincount(over_idx)
avg_runs_per_over =  (runs_per_over/balls_per_over)*6

task5_result = list(zip(unique_overs, avg_runs_per_over))
print("Total Runs Scored in Each Over")
print("-" * 27)

for over_num, runs in task5_result:
    print(f"Over {over_num:<3}            {runs:.2f} ")
```

Task 6: Boundary Analysis

```python
fours=batsman_runs==4
sixes=batsman_runs==6

print(f"Total Fours: {np.sum(fours)}")
print(f"Total Sixes: {np.sum(sixes)}")  

name_map = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Rising Pune Supergiant': 'Rising Pune Supergiants'
}

teams = np.array([name_map.get(t, t) for t in batting_team])

boundary_mask = fours | sixes
boundary_teams = teams[boundary_mask]
unique_teams, teams_idx = np.unique(boundary_teams, return_inverse=True)
boundary_counts = np.bincount(teams_idx)

max_index = np.argmax(boundary_counts)
top_boundary_team = unique_teams[max_index]
max_boundaries = boundary_counts[max_index]

print("Team with most boundaries:", top_boundary_team)
print("Total boundaries by this team:", max_boundaries)
```

Task 7: Death Overs Analysis

```python
death_mask = over > 15
total_death_runs = np.sum(total_runs[death_mask])
print("Total runs scored in death overs (15+):", total_death_runs)

death_over_teams = teams[death_mask]
unique_death_teams, death_teams_idx = np.unique(death_over_teams, return_inverse=True)
death_runs_per_team = np.bincount(death_teams_idx, weights=total_runs[death_mask])

max_index=np.argmax(death_runs_per_team)
top_death_team = unique_death_teams[max_index]
print("Team with most runs in death overs:", top_death_team)
print("Total runs scored by this team in death overs:", int(death_runs_per_team[max_index]))
```

Task 8: Highest Scoring Match

```python
match_runs = np.bincount(match_idx, weights=total_runs)
max_index=np.argmax(match_runs)
top_match_id = unique_matches[max_index]
print("Match with highest total runs:", int(top_match_id)) 
print("Total runs in that match:", int(match_runs[max_index]))
```

Task 9: Runs per Team per Match

```python
teams_list = np.column_stack((match_id, batting_team))

unique_list_of_teams, list_of_teams_idx = np.unique(
    teams_list, axis=0, return_inverse=True
)

runs_per_team = np.zeros(len(unique_list_of_teams))

np.add.at(runs_per_team, list_of_teams_idx, total_runs)

match_ids_unique = unique_list_of_teams[:, 0]
teams_unique = unique_list_of_teams[:, 1]

task9_result = list(zip(match_ids_unique, teams_unique, runs_per_team))

print(" Total Runs Scored by Each Team in Each Match")
print("-" * 46)

for mid, team, runs in task9_result:
    print(f"{int(mid)} - {team:<27} {int(runs)} runs")
```

Task 10: Match Winner Approximation

```python
winner_team = np.empty(len(unique_matches), dtype=object)
for i in range(len(unique_matches)):
    match_mask =match_idx ==i
    match_runs = total_runs[match_mask]
    match_teams = teams[match_mask]

    winner_team[i] = match_teams[np.argmax(match_runs)]
    task10_result = list(zip(unique_matches, winner_team))

print("Match Winners")
print("-" * 30)

for mid, team in task10_result:
    print(f"{int(mid):<10} {team}")
```

Task 11: Match Scorecard Generation

```python
match_teams = np.column_stack((match_id, batting_team))
unique_match_teams, idx, match_teams_idx = np.unique(match_teams, axis=0, return_index=True, return_inverse=True)


order = np.argsort(idx)
unique_match_teams = unique_match_teams[order]


inverse_order = np.empty_like(order)
inverse_order[order] = np.arange(len(order))
match_teams_idx = inverse_order[match_teams_idx]


runs=np.zeros(len(unique_match_teams))
np.add.at(runs, match_teams_idx, total_runs)

match_ids_unique = unique_match_teams[:, 0]
teams_unique = unique_match_teams[:, 1]

printed_matches = set()

print("Match Scorecards")
print("-" * 40)


for mid, team, r in zip(match_ids_unique, teams_unique, runs):
    mid_int = int(mid)

    if mid_int not in printed_matches:
        print(f"\nMatch {mid_int}:")
        printed_matches.add(mid_int)

    print(f"  {team:<27} {int(r)} runs")
```
